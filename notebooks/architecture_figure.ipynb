# Architecture figure for the report

Draws the **default** hybrid pipeline from `ARCHITECTURE.md`:
forecast (XGBoost alpha) → risk adjustment → alpha rule + PPO residual → sizing, constraints, reward.

Run all cells. Exports go to `notebooks/figures/` (`pdf`/`svg` for LaTeX or Word, `png` at 300 dpi).

Suggested caption:

> **Figure.** Hybrid trading architecture (default). XGBoost predicts 20-day active returns. A z-score rule proposes Buy/Hold/Sell; PPO outputs a residual `{DOWN, KEEP, UP}` around that rule. Position size is $\Delta w = a \cdot B_t \cdot q$, then long-only constraints and transaction costs. The reward is log-return minus a turnover penalty plus an alpha-alignment bonus.

````python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "ARCHITECTURE.md").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "notebooks"))

from architecture_draw import draw_architecture, draw_architecture_overview

OUT_DIR = ROOT / "notebooks" / "figures"
OUT_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update(
    {
        "font.family": "DejaVu Sans",
        "font.size": 11,
        "pdf.fonttype": 42,
        "ps.fonttype": 42,
        "savefig.bbox": "tight",
        "savefig.facecolor": "white",
    }
)
print(f"Saving figures to {OUT_DIR}")
````

````python
fig, ax = plt.subplots(figsize=(9.6, 13.6))
fig.patch.set_facecolor("white")
ax.set_facecolor("white")
draw_architecture(ax)

for ext in ("pdf", "svg", "png"):
    path = OUT_DIR / f"architecture.{ext}"
    fig.savefig(path, dpi=300, pad_inches=0.12)
    print("wrote", path)

try:
    get_ipython()
    plt.show()
except NameError:
    plt.close(fig)
````
